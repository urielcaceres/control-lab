## Sistema de Control de Acceso al Laboratorio Informatico

```python
#definir contadores
permitido=0
denegado=0
total=0
continuar= "si"


while continuar== "si":
    print("=====Nueva Solicitud=====")
    #pedir datos
    nombre= input("Ingresar nombre del Estudiante:")
    matricula= input("Esta el estudiante matriculado? si/no").lower()
    credencial= input("Tiene credencial valida? si/no").lower()
    edad= int(input("Ingrese su edad:"))
    total= total +1

    if not (matricula=="si"):
        print("El usuario no esta matriculado")
    if not (credencial=="si"):
        print("No tiene credencial valida")
    if not (edad>=18):
        print("Es menor de edad")
    if (matricula !="si" and matricula !="no") or (credencial!="si" and credencial!="no"):
        print("Incorrecto! Solo se debe ingresar si/no")
    elif matricula =="si" and credencial =="si" and edad >=18:
        print("Acceso permitido!",nombre, "puede ingresar al laboratorio")
        permitido= permitido+1
    else:
        print("Acceso denegado")
        denegado= denegado+1
    continuar= input("Desea continuar? si/no").lower()
    if total > 0:
        porcentaje= permitido / total * 100
    else:
        porcentaje= 0
print("=====Resumen=====")
print("Solicitudes procesadas",total)
print("Accesos permitidos",permitido)
print("Accesos denegados",denegado)
print("El porcentaje de aprobados es",porcentaje,"%")
```
